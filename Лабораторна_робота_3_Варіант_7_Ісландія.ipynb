[cite_start][Посилання на Вікіпедію про прапор](https://uk.wikipedia.org/wiki/%D0%9F%D1%80%D0%B0%D0%BF%D0%BE%D1%80_%D0%86%D1%81%D0%BB%D0%B0%D0%BD%D0%B4%D1%96%D1%97) [cite: 19]

```python
# --- Допоміжні функції для переведення ---

# Переведення цілої частини (діленням на 2)
def int_to_bin(integer):
    if integer == 0:
        return '0'
    binary = ''
    while integer > 0:
        binary = str(integer % 2) + binary
        integer //= 2
    return binary

# Переведення дробової частини (множенням на 2)
def frac_to_bin(fraction, precision=10):
    if fraction == 0:
        return ''
    binary = ''
    for _ in range(precision):
        if fraction == 0:
            break
        fraction *= 2
        bit = int(fraction)
        binary += str(bit)
        fraction -= bit
    return '.' + binary

# --- Основна функція task_7() ---
def task_7():
    """
    Реалізує переведення числа (цілого або з рухомою комою)
    з десяткової системи числення в двійкову з контролем введення.
    """
    print("Введіть число (ціле або дробове): ")

    # 1. Контроль введення (використовуємо input() для Colab/Jupyter)
    decimal_str = input().strip()

    if not decimal_str:
        return "Помилка. Введення не може бути порожнім."

    try:
        # Спроба перетворення на число з плаваючою комою для валідації
        number = float(decimal_str)
    except ValueError:
        # [cite_start]Обробка помилки при нечисловому введенні [cite: 25]
        return f"Помилка. У списку присутнє нечислове значення: '{decimal_str}'"


    # 2. Визначення знака
    sign = '0'
    if number < 0:
        sign = '1'
        abs_number = abs(number)
    else:
        abs_number = number

    # 3. Розділення на цілу та дробову частини
    integer_part = int(abs_number)
    fractional_part = abs_number - integer_part

    # 4. Переведення цілої частини
    binary_integer = int_to_bin(integer_part)

    # 5. Переведення дробової частини
    binary_fraction = frac_to_bin(fractional_part)

    # Формат виводу: знак|ціла частина.дробова частина
    binary_result = f"{sign}|{binary_integer}{binary_fraction}"

    # [cite_start]Вивід відповідно до прикладів [cite: 22, 23, 24, 25, 29, 31]
    return f"'{decimal_str} у двійковій системі буде {binary_result}'"
```
